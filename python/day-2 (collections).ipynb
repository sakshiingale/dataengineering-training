```python
sample_list1 = tuple(""" twinkle twinkle little star, 
how i wonder what you are,
up above the world so high,
like a diamond in the sky""".split())
print(sample_list1)
```

```python
sample_list1 = tuple(""" twinkle twinkle little star, 
how i wonder what you are,
up above the world so high,
like a diamond in the sky""".split())

# Accessing elements
first_elem = sample_list1[0]
last_elem = sample_list1[-1]

# Slicing
slice_tuple = sample_list1[1:4]

# Length
tuple_length = len(sample_list1)

# Count occurrences
count_twinkle = sample_list1.count('twinkle')

# Find index
index_little = sample_list1.index('little')

# Concatenation
concat_tuple = sample_list1 + ('extra',)

# Repetition
repeat_tuple = sample_list1 * 2

# Membership
is_star_present = 'star,' in sample_list1

# Iteration
for word in sample_list1:
    pass

# Min/Max (if all elements are comparable)
try:
    min_word = min(sample_list1)
    max_word = max(sample_list1)
except TypeError:
    min_word = None
    max_word = None

# Conversion to list
tuple_as_list = list(sample_list1)

# Display results
display({
    "first_elem": first_elem,
    "last_elem": last_elem,
    "slice_tuple": slice_tuple,
    "tuple_length": tuple_length,
    "count_twinkle": count_twinkle,
    "index_little": index_little,
    "concat_tuple": concat_tuple,
    "repeat_tuple": repeat_tuple,
    "is_star_present": is_star_present,
    "min_word": min_word,
    "max_word": max_word,
    "tuple_as_list": tuple_as_list
})
```

```python
# Dictionary examples
dict1 = {"name": "Alice", "age": 30, "city": "Pune"}
dict2 = dict([("company", "Zensar"), ("location", "pune"), ("employees", 10000)])
dict3 = {1: "one", 2: "two", 3: "three"}

# Accessing values
name = dict1["name"]
brand = dict2.get("brand")

# Adding/updating entries
dict1["salary"] = 50000
dict2["location"] = "Mumbai"

# Removing entries
removed_age = dict1.pop("age")
del dict3[2]

# Keys, values, items
keys = list(dict1.keys())
values = list(dict1.values())
items = list(dict2.items())

# Membership
is_city_present = "city" in dict1

# Iteration
for k, v in dict1.items():
    pass

# Merging dictionaries
merged_dict = {**dict1, **dict2}

# Clearing dictionary
dict3.clear()

# Display results
display({
    "dict1": dict1,
    "dict2": dict2,
    "dict3": dict3,
    "name": name,
    "brand": brand,
    "removed_age": removed_age,
    "keys": keys,
    "values": values,
    "items": items,
    "is_city_present": is_city_present,
    "merged_dict": merged_dict
})
```

```python
file_path = input("enter file path:") 
poem_str = dbutils.fs.head(file_path, 10000)

# Create a list of words from the string
words_list = poem_str.split()

# Count frequency of each word
from collections import Counter
word_freq = Counter(words_list)

# Display the word frequency dictionary
display(dict(word_freq))

# Write output to output.txt in the same directory
output_dir = "/".join(file_path.split("/")[:-1])
output_path = f"{output_dir}/output.txt"
output_content = "\n".join([f"{word}: {count}" for word, count in word_freq.items()])
dbutils.fs.put(output_path, output_content, overwrite=True)
```
