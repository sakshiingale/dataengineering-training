```python
def calculator():
    while True:
        print("Select operation:")
        print("1. Add")
        print("2. Subtract")
        print("3. Multiply")
        print("4. Divide")
        print("5. Power")
        print("6. Exit")
        choice = input("Enter choice (1-6): ")
        
        if choice == '6':
            print("Exiting calculator.")
            break
        
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
        
        if choice == '1':
            result = num1 + num2
            print(f"Result: {result}")
        elif choice == '2':
            result = num1 - num2
            print(f"Result: {result}")
        elif choice == '3':
            result = num1 * num2
            print(f"Result: {result}")
        elif choice == '4':
            if num2 != 0:
                result = num1 / num2
                print(f"Result: {result}")
            else:
                print("Error: Division by zero.")
        elif choice == '5':
            result = num1 ** num2
            print(f"Result: {result}")
        else:
            print("Invalid choice.")

calculator()
```

```python
def rename_file():
    file_path = input("Enter the current file path: ")
    new_name = input("Enter the new file name: ")
    extension = file_path.split('.')[-1] if '.' in file_path else ''
    parent_dir = "/".join(file_path.split("/")[:-1])
    if extension and not new_name.endswith(f".{extension}"):
        new_name = f"{new_name}.{extension}"
    new_path = f"{parent_dir}/{new_name}"
    dbutils.fs.mv(file_path, new_path)
    print("File renamed successfully")
rename_file()
```

```python
import os

def add_file_with_count(src_file, dst_dir):
    files_before = [f for f in dbutils.fs.ls(dst_dir)]
    count_before = len(files_before)
    dbutils.fs.cp(src_file, dst_dir)
    files_after = [f for f in dbutils.fs.ls(dst_dir)]
    count_after = len(files_after)
    print(f"Number of files before: {count_before}")
    print(f"Number of files after: {count_after}")
    original_name = os.path.basename(src_file)
    new_name = f"{count_after}_{original_name}"
    dst_path = os.path.join(dst_dir, new_name)
    dbutils.fs.mv(f"{dst_dir}/{original_name}", dst_path)
    print(f"File copied and renamed to: {dst_path}")

src_file = input("Enter the path of the file to copy: ")
dst_dir = input("Enter the destination directory path: ")
add_file_with_count(src_file, dst_dir)
```

```python
import time

def add_file_with_timestamp(src_file, dst_dir):
    original_name = os.path.basename(src_file)
    timestamp = time.strftime("%Y%m%d%H%M%S")
    new_name = f"{timestamp}_{original_name}"
    dst_path = os.path.join(dst_dir, new_name)
    dbutils.fs.mv(src_file, dst_path)
    print(f"File moved and renamed to: {dst_path}")

src_file = input("Enter the path of the file to move: ")
dst_dir = input("Enter the destination directory path: ")
add_file_with_timestamp(src_file, dst_dir)
```

```python
source_path = input("Enter source file path: ")
volume_dir = input("Enter volume directory: ")
file_name = source_path.split('/')[-1]
if file_name.endswith('.csv'):
    dest_dir = f"{volume_dir}/csv_files"
elif file_name.endswith('.json'):
    dest_dir = f"{volume_dir}/json_files"
else:
    print("Unsupported file type.")

dbutils.fs.mkdirs(dest_dir)
destination_path = f"{dest_dir}/{file_name}"
dbutils.fs.cp(source_path, destination_path)

# Read first value from file
first_value = None
extension = file_name.split('.')[-1]
if extension == 'csv':
    import pandas as pd
    import io
    with open(source_path, 'r') as f:
        df = pd.read_csv(f, nrows=1)
        first_value = str(df.iloc[0, 0])
elif extension == 'json':
    import json
    with open(source_path, 'r') as f:
        data = json.load(f)
        if isinstance(data, list) and len(data) > 0:
            first_value = str(list(data[0].values())[0])
        elif isinstance(data, dict):
            first_value = str(list(data.values())[0])

if first_value:
    new_file_name = f"{first_value}.{extension}"
    new_destination_path = f"{dest_dir}/{new_file_name}"
    dbutils.fs.mv(destination_path, new_destination_path)
    print(f"File copied and renamed to: {new_destination_path}")
else:
    print(f"File copied to: {destination_path} (could not extract first value for renaming)")
```
