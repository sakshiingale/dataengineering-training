```python
# lets define input data path / files path 
# and 
# checkpoint dir path
```

```python
input_path = "/Volumes/dea_databricks_catalog/default/dea_volume/yellowtaxi_raw/csv/"

checkpoint_path = "/Volumes/dea_databricks_catalog/default/dea_volume/yellowtaxi_raw/checkpoints_dir/"

# lets define a new table to create and write the streaming output into it

table_name = "yt_raw_autoloader_str_output"
```

```python
# 1 - Read data from storage location (input_path) using autoloader
# 2 - Write data into table (yt_raw_autoloader_str_output)
```

```python
# 1 - Read data from storage location (input_path) using autoloader
# 2 - Write data into table (yt_raw_autoloader_str_output)
```

```python
yt_raw_str_DF = (
  spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option("cloudFiles.schemaLocation", checkpoint_path)
    .load(input_path))
```

```python
yt_raw_str_DF.isStreaming
```

```python
yt_raw_str_DF = (
  spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option("cloudFiles.schemaLocation", checkpoint_path)
    .load(input_path))

yt_raw_str_DF.writeStream.option("checkpointLocation", checkpoint_path).trigger(continuous='1 second').toTable(table_name)
```

```python
query = (
    yt_raw_str_DF.writeStream
    .option("checkpointLocation", checkpoint_path)
    .trigger(availableNow=True)
    .toTable(table_name)
)

query.awaitTermination()
```

```python
%sql
create or replace table yt_silver_str_delta as
select *
from (
  select
    TRY_CAST(passenger_count AS INT) as passenger_count,
    TRY_CAST(trip_distance AS DOUBLE) as trip_distance,
    TRY_CAST(total_amount AS DOUBLE) as total_amount,
    TRY_CAST(fare_amount AS DOUBLE) as fare_amount,
    tpep_pickup_datetime,
    tpep_dropoff_datetime,
    VendorID
    -- add other columns as needed, ensuring no duplicates
  from yt_raw_autoloader_str_output
)
where 
  passenger_count > 0
  and trip_distance > 0
  and total_amount > 0
  and fare_amount > 0
  and tpep_pickup_datetime != tpep_dropoff_datetime
  and VendorID is not null
```

```python
%sql
select count(*) from yt_silver_str_delta
```
