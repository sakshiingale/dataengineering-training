```sql
create table if not exists bronze_date
as
select *,
       year(tpep_pickup_datetime) as trip_year,
       month(tpep_pickup_datetime) as trip_month
from ldp_db.yt_bronze
```

```sql
create or replace function trip_year_month(dt timestamp)
returns struct<trip_year:int, trip_month:int>
return struct(year(dt) as trip_year, month(dt) as trip_month);
-- returns table<trip_year:int, trip_month:int>
-- return select year(dt) as trip_year, month(dt) as trip_month;

create table if not exists bronze_date1
as
select *,
       trip_year_month(tpep_pickup_datetime).trip_year as trip_year,
       trip_year_month(tpep_pickup_datetime).trip_month as trip_month
from ldp_db.yt_bronze
```

```sql
create or replace function mask_email(email string)
returns string
return translate(
  lower(email),
  'abcdefghijklmnopqrstuvwxyz0123456789',
  'bcdefghijklmnopqrstuvwxyza2345678901'
);

select customer_id, customer_name, mask_email(email) as masked_email, city
from customers_demo;
```

```sql
create or replace function mask_email_custom(email string)
returns string
return concat_ws('', transform(split(email, ''), c ->
  case
    when c rlike '[a-z]' then char(((ascii(c) - ascii('a') + 1) % 26) + ascii('a'))
    when c rlike '[A-Z]' then char(((ascii(c) - ascii('A') + 1) % 26) + ascii('A'))
    when c rlike '[0-9]' then cast(((cast(c as int) + 1) % 10) as string)
    else c
  end
));

select customer_id, customer_name, mask_email_custom(email) as masked_email, city
from customers_demo;
```
