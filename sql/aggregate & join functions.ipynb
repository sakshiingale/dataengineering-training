```sql
-- SUM: Total sales amount for each product
SELECT product_id, SUM(amount) AS total_sales
FROM sales
GROUP BY product_id;

-- COUNT: Number of sales per customer, only for customers older than 30
SELECT customer_id, COUNT(*) AS sales_count
FROM sales
WHERE customer_id IN (SELECT customer_id FROM customers WHERE age > 30)
GROUP BY customer_id;

-- AVG: Average sale amount per day, only for days with more than one sale
SELECT sale_date, AVG(amount) AS avg_amount
FROM sales
GROUP BY sale_date
HAVING COUNT(*) > 1;

-- MIN: Minimum sale amount per product, only for products in 'Electronics' category
SELECT s.product_id, MIN(s.amount) AS min_amount
FROM sales s
JOIN products p ON s.product_id = p.product_id
WHERE p.category = 'Electronics'
GROUP BY s.product_id;

-- MAX: Maximum sale amount per city, only for cities with at least 2 sales
SELECT c.city, MAX(s.amount) AS max_amount
FROM sales s
JOIN customers c ON s.customer_id = c.customer_id
GROUP BY c.city
HAVING COUNT(*) >= 2;

-- GROUP BY with multiple aggregates and HAVING/WHERE
SELECT p.category, COUNT(s.sale_id) AS num_sales, SUM(s.amount) AS total_amount, AVG(s.amount) AS avg_amount
FROM sales s
JOIN products p ON s.product_id = p.product_id
WHERE s.amount > 150
GROUP BY p.category
HAVING SUM(s.amount) > 500;

-- STDDEV: Standard deviation of sale amounts per product
SELECT product_id, STDDEV(amount) AS stddev_amount
FROM sales
GROUP BY product_id;

-- VARIANCE: Variance of sale amounts per customer, only for customers in 'Texas' cities
SELECT s.customer_id, VARIANCE(s.amount) AS variance_amount
FROM sales s
JOIN customers c ON s.customer_id = c.customer_id
WHERE c.city IN ('Houston', 'Dallas', 'Austin', 'San Antonio', 'Fort Worth')
GROUP BY s.customer_id;

-- COUNT DISTINCT: Number of distinct products sold per city, only for 
-- cities with more than 2 customers
SELECT c.city, COUNT(DISTINCT s.product_id) AS distinct_products
FROM sales s
JOIN customers c ON s.customer_id = c.customer_id
GROUP BY c.city
HAVING COUNT(DISTINCT c.customer_id) > 2;
```

```sql
-- INNER JOIN: transactions with matching accounts and customers
SELECT t.transaction_id, t.amount, a.account_type, c.customer_name
FROM transactions t
INNER JOIN accounts a ON t.account_id = a.account_id
INNER JOIN customers c ON t.customer_id = c.customer_id;
```

```sql
-- LEFT JOIN: All transactions, show NULLs for accounts/customers not matched
SELECT t.transaction_id, t.amount, a.account_type, c.customer_name
FROM transactions t
LEFT JOIN accounts a ON t.account_id = a.account_id
LEFT JOIN customers c ON t.customer_id = c.customer_id;
```

```sql

-- RIGHT JOIN: All accounts/customers, show NULLs for transactions not matched
SELECT t.transaction_id, t.amount, a.account_type, c.customer_name
FROM transactions t
RIGHT JOIN accounts a ON t.account_id = a.account_id
RIGHT JOIN customers c ON t.customer_id = c.customer_id;
```

```sql
-- FULL OUTER JOIN: All records from transactions, accounts, customers
SELECT t.transaction_id, t.amount, a.account_type, c.customer_name
FROM transactions t
FULL OUTER JOIN accounts a ON t.account_id = a.account_id
FULL OUTER JOIN customers c ON t.customer_id = c.customer_id;
```

```sql
-- CROSS JOIN: All combinations of accounts and customers
SELECT a.account_id, a.account_type, c.customer_id, c.customer_name
FROM accounts a
CROSS JOIN customers c;
```
