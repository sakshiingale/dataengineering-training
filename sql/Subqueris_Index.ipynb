```sql
select emp_id, dept_id, emp_name , salary, 
RANK() OVER(order by salary desc) AS emp_rank from employees
```

```sql
select e1.emp_id, e1.dept_id, e1.emp_name, e1.salary,
  1 + (select count(*) from employees e2 where e2.salary > e1.salary) as emp_rank
from employees e1
```

```sql
select e1.emp_id, e1.dept_id, e1.emp_name, e1.salary,
  1 + (select count(*) from employees e2 where e2.salary > e1.salary) as emp_rank
from employees e1 order by emp_rank
```

```sql
select e1.emp_id, e1.dept_id, e1.emp_name, e1.salary,
  1 + (select count(*) from employees e2 
        where e2.dept_id = e1.dept_id and e2.salary > e1.salary) as emp_rank 
from employees e1 order by emp_rank
```
