```sql
select  emp_id, emp_name, salary,
row_number() over(order by emp_id) as Row_Num,
RANK() OVER(order by salary desc) AS emp_rank,
DENSE_RANK() OVER(order by salary desc) AS emp_dense_rank from employees
```

```sql
select  emp_id, emp_name, salary,
row_number() over(order by emp_id) as Row_Num,
RANK() OVER(order by salary desc) AS emp_rank,
DENSE_RANK() OVER(order by salary desc) AS emp_dense_rank from employees order by emp_rank, emp_dense_rank
```

```sql
create table emp_analytics
as
select  emp_id, emp_name, salary,
row_number() over(order by emp_id) as Row_Num,
RANK() OVER(order by salary desc) AS emp_rank,
DENSE_RANK() OVER(order by salary desc) AS emp_dense_rank from employees order by emp_rank, emp_dense_rank
```

```sql
select * from emp_analytics
```
